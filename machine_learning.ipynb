# Random Forest

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, classification_report


eth = pd.read_csv("project_3/data/ethiopia_final.csv")
synth = pd.read_csv("project_3/data/synthetic_final.csv")
ind = pd.read_csv("project_3/data/india_final.csv")

# same features and target for all datasets
features = ["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]
target = "label"
results = []

eth.columns = eth.columns.str.strip()
eth["label"] = eth["label"].astype(str).str.strip().str.lower()
eth = eth.dropna(subset=features + [target])
X = eth[features]
y = eth[target]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13, stratify=y)
model= RandomForestClassifier(n_estimators=200, random_state=13, n_jobs=-1)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
weighted_f1 = f1_score(y_test, y_pred, average="weighted")
macro_f1 = f1_score(y_test, y_pred, average="macro")
accuracy = accuracy_score(y_test, y_pred)
weighted_f1 = f1_score(y_test, y_pred, average="weighted")
macro_f1 = f1_score(y_test, y_pred, average="macro")

print("ethipia dataset")
print(f"the accuracy of the model is: {accuracy:.4f}")
print(f"the weighted F1 score of the model is: {weighted_f1:.4f}")
print(f"the macro F1 score of the model is: {macro_f1:.4f}")
```

# SVM

```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, classification_report

eth = pd.read_csv("project_3/data/ethiopia_final.csv")
synth = pd.read_csv("project_3/data/synthetic_final.csv")
ind = pd.read_csv("project_3/data/india_final.csv")
features = ["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]
target = "label"
eth.columns = eth.columns.str.strip()
eth["label"] = eth["label"].astype(str).str.strip().str.lower()
eth = eth.dropna(subset=features + [target])
X = eth[features]
y = eth[target]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13, stratify=y)
model = make_pipeline(StandardScaler(), SVC(kernel="rbf", C=1.0, gamma="scale", random_state=13))
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
weighted_f1 = f1_score(y_test, y_pred, average="weighted")
macro_f1 = f1_score(y_test, y_pred, average="macro")

print("ethiopia dataset")
print(f"accuracy is {accuracy:.4f}")
print(f"weighted F1 is {weighted_f1:.4f}")
print(f"macro F1 is {macro_f1:.4f}")

print("classification report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

# KNN

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, classification_report

eth = pd.read_csv("project_3/data/ethiopia_final.csv")
synth = pd.read_csv("project_3/data/synthetic_final.csv")
ind = pd.read_csv("project_3/data/india_final.csv")
features = ["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]
target = "label"
eth.columns = eth.columns.str.strip()
eth["label"] = eth["label"].astype(str).str.strip().str.lower()
eth = eth.dropna(subset=features + [target])
X = eth[features]
y = eth[target]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13, stratify=y)
model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
weighted_f1 = f1_score(y_test, y_pred, average="weighted")
macro_f1 = f1_score(y_test, y_pred, average="macro")

print("ethiopia dataset")
print(f"accuracy is {accuracy:.4f}")
print(f"weighted F1 is {weighted_f1:.4f}")
print(f"macro F1 is {macro_f1:.4f}")

print("classification report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

# Logistic Regression

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, classification_report

eth = pd.read_csv("project_3/data/ethiopia_final.csv")
synth = pd.read_csv("project_3/data/synthetic_final.csv")
ind = pd.read_csv("project_3/data/india_final.csv")
features = ["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]
target = "label"
eth.columns = eth.columns.str.strip()
eth["label"] = eth["label"].astype(str).str.strip().str.lower()
eth = eth.dropna(subset=features + [target])
X = eth[features]
y = eth[target]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13, stratify=y)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000, random_state=13))
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
weighted_f1 = f1_score(y_test, y_pred, average="weighted")
macro_f1 = f1_score(y_test, y_pred, average="macro")

print("ethiopia dataset")
print(f"accuracy is {accuracy:.4f}")
print(f"weighted F1 is {weighted_f1:.4f}")
print(f"macro F1 is {macro_f1:.4f}")

print("classification report:")
print(classification_report(y_test, y_pred, zero_division=0))
```
