# Random Forest

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, classification_report


eth = pd.read_csv("project_3/data/ethiopia_final.csv")
synth = pd.read_csv("project_3/data/synthetic_final.csv")
ind = pd.read_csv("project_3/data/india_final.csv")

# same features and target for all datasets
features = ["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]
target = "label"
results = []

eth.columns = eth.columns.str.strip()
eth["label"] = eth["label"].astype(str).str.strip().str.lower()
eth = eth.dropna(subset=features + [target])
X = eth[features]
y = eth[target]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13, stratify=y)
model= RandomForestClassifier(n_estimators=200, random_state=13, n_jobs=-1)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
weighted_f1 = f1_score(y_test, y_pred, average="weighted")
macro_f1 = f1_score(y_test, y_pred, average="macro")
accuracy = accuracy_score(y_test, y_pred)
weighted_f1 = f1_score(y_test, y_pred, average="weighted")
macro_f1 = f1_score(y_test, y_pred, average="macro")

print("ethipia dataset")
print(f"the accuracy of the model is: {accuracy:.4f}")
print(f"the weighted F1 score of the model is: {weighted_f1:.4f}")
print(f"the macro F1 score of the model is: {macro_f1:.4f}")
```
