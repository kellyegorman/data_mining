```python
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
eth = pd.read_csv("project_3/data/ethiopia_final.csv")
features = ["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]
target = "label"

eth.columns = eth.columns.str.strip()
eth["label"] = eth["label"].astype(str).str.strip().str.lower()
eth = eth.dropna(subset=features + [target])

X = eth[features]
y = eth[target]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=13, stratify=y
)

feature_sets = {
    "all features": features,
    "soil only": ["N", "P", "K", "ph"],
    "climate only": ["temperature", "humidity", "rainfall"],
    "N pH temperature": ["N", "ph", "temperature"],
    "soil and temperature": ["N", "P", "K", "ph", "temperature"]
}

models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=15,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=13,
        n_jobs=-1
    ),
    "Extra Trees": ExtraTreesClassifier(
        n_estimators=300,
        max_depth=15,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=13,
        n_jobs=-1
    )
}

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=13)
results = []

for model_name, model in models.items():
    for feature_name, selected_features in feature_sets.items():
        scores = cross_validate(
            model,
            X_train[selected_features],
            y_train,
            cv=cv,
            scoring={"accuracy": "accuracy", "macro_f1": "f1_macro"},
            n_jobs=-1
        )

        results.append({
            "model": model_name,
            "features": feature_name,
            "cv_accuracy": scores["test_accuracy"].mean(),
            "cv_macro_f1": scores["test_macro_f1"].mean()
        })

results_df = pd.DataFrame(results)
results_df = results_df.sort_values("cv_macro_f1", ascending=False)

print(results_df.to_string(index=False))

best = results_df.iloc[0]
best_model = models[best["model"]]
best_features = feature_sets[best["features"]]

best_model.fit(X_train[best_features], y_train)
y_pred = best_model.predict(X_test[best_features])

print("\nbest model:", best["model"])
print("best features:", best_features)
print(f"test accuracy is {accuracy_score(y_test, y_pred):.4f}")
print(f"weighted F1 is {f1_score(y_test, y_pred, average='weighted'):.4f}")
print(f"macro F1 is {f1_score(y_test, y_pred, average='macro'):.4f}")

print("\nclassification report:")
print(classification_report(y_test, y_pred, zero_division=0))
```
