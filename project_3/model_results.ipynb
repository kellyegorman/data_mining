```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
results = pd.read_csv("project_3/data/model_results_summary.csv")
display(results)
```

# Ethiopia Results 

```python
eth_results = results[results["Dataset"] == "Ethiopia"]

plt.figure(figsize = (10,5))
sns.barplot(data = eth_results, x = "Model", y = "Macro F1")
plt.title("Ethiopia Results by Macro F1")
plt.xticks(rotation=35, ha="right")
```

# India Results

```python
in_results = results[results["Dataset"] == "India"]

plt.figure(figsize = (10,5))
sns.barplot(data = in_results, x = "Model", y = "Macro F1")
plt.title("India Results by Macro F1")
plt.xticks(rotation=35, ha="right")
```

# Ethiopia vs. India Comparison 

```python
heatmap_data = results.pivot(
    index="Model",
    columns="Dataset",
    values="Macro F1"
)

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".3f",
    cmap="YlGnBu",
    vmin=0,
    vmax=1
)

plt.title("Model Performance Across Datasets")
plt.xlabel("Dataset")
plt.ylabel("Model")
plt.tight_layout()
plt.show()
```
