```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(0, 1, (500, 50))
from scipy.spatial.distance import pdist
i = 2
distances = pdist(data[:, :i], metric='euclidean')
print(distances)
Max_dist = np.max(distances)
Min_dist = np.min(distances)
value = np.log10((Max_dist - Min_dist) / Min_dist)
```

```python
values = []

for i in range(2, 51):
    current_data = data[:, :i]
    distances = pdist(current_data, metric='euclidean')
    Max_dist = np.max(distances)
    Min_dist = np.min(distances)
    value = np.log10((Max_dist - Min_dist) / Min_dist)
    values.append(value)
```

```python
plt.plot(range(2, 51), values)
plt.xlabel("Number of dimensions")
plt.ylabel("log10((MAX_DIST - MIN_DIST) / MIN_DIST)")
plt.grid(True)
plt.show()
```
