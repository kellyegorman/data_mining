```python
import pandas as pd
import matplotlib.pyplot as plt

housing = pd.read_csv("housing.csv")
print(housing.head())
print(housing.info())
print(housing.describe())
```

```python
housing.hist(figsize=(15, 12))
plt.tight_layout()
plt.show()
```

```python
housing.boxplot(figsize=(15, 8))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
