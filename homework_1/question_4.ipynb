# Part 1 

```python
import pandas as pd
import matplotlib.pyplot as plt

housing = pd.read_csv("housing.csv")
print(housing.head())
print(housing.info())
print(housing.describe())
```

# Part 2

```python
housing.hist(figsize=(15, 12))
plt.tight_layout()
plt.show()
```

```python
housing.boxplot(figsize=(15, 8))
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Part 3

```python
plt.scatter(housing["longitude"], housing["latitude"])

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("California Housing Locations")

plt.show()
```

```python
plt.scatter(
    housing["longitude"],
    housing["latitude"],
    s=housing["population"] / 100,
    alpha=0.4
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("California Housing Locations")

plt.show()
```

```python
plt.scatter(
    housing["longitude"],
    housing["latitude"],
    s=housing["population"] / 100,
    c=housing["median_house_value"],
    alpha=0.4
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("California Housing Locations")
plt.colorbar(label="Median House Value")

plt.show()
```

# Part 4

```python
plt.scatter(
    housing["median_income"],
    housing["median_house_value"]
)

plt.xlabel("Median Income")
plt.ylabel("Median House Value")
plt.title("Median Income vs. Median House Value")

plt.show()
```

```python
correlation = housing["median_income"].corr(
    housing["median_house_value"]
)

print("Correlation coefficient:", correlation)
```
